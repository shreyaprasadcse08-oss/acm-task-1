```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load clean dataset
url = 'https://raw.githubusercontent.com/spscientist/students-performance-in-exams/master/StudentsPerformance.csv'
df = pd.read_csv(url)
```

```python
# 1. Histogram: Distribution of Math Scores
plt.figure(figsize=(8, 5))
sns.histplot(df['math score'], kde=True, color='skyblue')
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Bar Chart: Average Scores by Test Preparation Course
plt.figure(figsize=(8, 5))
df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean().plot(kind='bar', figsize=(8,5))
plt.title('Average Scores by Test Preparation Course')
plt.xlabel('Test Preparation Course')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.show()
```

```python
# 3. Box Plot: Math Score by Parental Level of Education
plt.figure(figsize=(10, 6))
sns.boxplot(x='parental level of education', y='math score', data=df)
plt.title('Math Score by Parental Level of Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Math Score')
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Scatter Plot: Reading Score vs. Writing Score
plt.figure(figsize=(8, 5))
sns.scatterplot(x='reading score', y='writing score', hue='gender', data=df)
plt.title('Reading Score vs. Writing Score')
plt.xlabel('Reading Score')
plt.ylabel('Writing Score')
plt.show()
```

```python
# 5. Heatmap: Score Correlations
plt.figure(figsize=(8, 5))
sns.heatmap(df[['math score', 'reading score', 'writing score']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Test Scores')
plt.show()
```
